```python
from shannonist.models.mlp import MultiMLP
import torch

model = MultiMLP(
    input_dim=32,
    count=4,
    output_dim=8,
    hidden_dim=(64, 64),
)
x = torch.randn(16, 10, 4, 32)
out = model(x)
out.shape
```

```python
import torch

from shannonist.models import InvertibleLinear

transform = InvertibleLinear(dim=4, bias=True)
x = torch.randn(8, 4)
forward = transform(x)
inverse = transform.inverse(forward.value)

{
    "input_shape": x.shape,
    "output_shape": forward.value.shape,
    "forward_log_abs_det": forward.log_abs_det[0].item(),
    "inverse_log_abs_det": inverse.log_abs_det[0].item(),
    "max_reconstruction_error": (x - inverse.value).abs().max().item(),
}
```
